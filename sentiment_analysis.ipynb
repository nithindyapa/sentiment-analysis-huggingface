```python
from transformers import pipeline
```

```python
classifier = pipeline("sentiment-analysis")
```

```python
text = "I love learning Artificial Intelligence!"
```

```python
result = classifier(text)

print(result)
```

```python
label = result[0]["label"]
score = result[0]["score"]

print("Input Text:", text)
print("Sentiment:", label)
print("Confidence:", round(score * 100, 2), "%")
```

```python
texts = [
    "I absolutely love this product. It works perfectly!",
    "The service was terrible. I will never buy again.",
    "The product arrived yesterday."
]

for text in texts:
    result = classifier(text)

    print("Text:", text)
    print("Sentiment:", result[0]["label"])
    print("Confidence:", round(result[0]["score"] * 100, 2), "%")
    print("-" * 50)
```
